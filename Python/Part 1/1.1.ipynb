### **Концепция переменных в Python**




В Python переменная - это идентификатор, который ссылается на некий объект в памяти компьютера.

Ключевые особенности Python:

- Динамическая типицазия, т.е. тип переменной будет определяться во время выполнения.
- Ссылочная модель, т.е. переменные хранят ссылки, а не сами значения
- Автоматическое освобождение памяти


```python
# Переменная "a" ссылается на объект "5"
a = 10 # Создается объект int со значением 5, "a" - ссылка на него
print(type(a))
```

### Динамическая типизация



Как уже упоминалось ранее динамическая типизация обозначает то, что тип переменной определяется в момент выполнения программы и может меняться в дальнейшем.

Плюсы динамической типизации:
- гибкость и скорость разработки
- меньший объём кода

Минусы динамической типизации:
- ошибки типов возникают только при выполнении
- сложнее читать другой код (неясно какой тип ожидать)
- меньшая производительность из-за проверок типов

```python
a = 10      # a - int
print(type(a))
a = "текст" # a - str (произошло изменения типа с числа на строку)
print(type(a))
```

**Функции для работы с типами**
> type()

Данная функция выводит тип объекта


```python
a = 10          # a - int
print(type(a))
a = "текст"     # a - str (произошло изменения типа с числа на строку)
print(type(a))
a = ["1", "2", "3"]# a - list (тип поменялся со строки на список)
print(type(a))
```

> isinstance()

Данная функция проверяет, является ли объект экземпляром указанного класса. (Вместо неё можно использовать "==", но функция предпочтительнее)

```python
a = 10          # a - int
print(type(a))
print(type(a) == int)
print(isinstance(a, int))
```

> id() 

Функция для получения идентификатора объекта в памяти

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a
print(id(a))
print(id(b))
print(id(c))
```

В данном случае мы увиди разные id для "a" и "b", это связано с тем, что при создании списка Python создаёт новый объект в памяти, однако для "c" и "a" id будут равны, так как ссылка будет такой же, так как создаётся не новый список, а лишь приравнивается к старому.

Если мы попробуем провернуть схожие действия с числами, то увидим следующее:

```python
a = 256
b = 256
print(id(a), id(b), id(a) == id(b))

c = 257
d = 257
print(id(c), id(d), id(c) == id(d))

e = -5
f = -5
print(id(e), id(f), id(e) == id(f))

g = -6
h = -6
print(id(g), id(h), id(g) == id(h))
```

Это связано с тем, что Python кэширует целые небольшие числа (от -5 до 256). 

То есть при создании переменной "a" равной 256 мы лишь ссылаемся на уже существующую ячейку памяти, что позволяет экономить память и время при создании объектов.

Также кэшированию будут подвергаться короткие строки без пробелов и некоторые пустые контейнеры.

```python
a = "abcd"
b = "abcd"
print(id(a) == id(b))
```

Строки длиной ≤ 4096 символов, состоящие только из ASCII символов (латинские буквы, цифры, нижнее подчёркивание), обычно кэшируются.

### Правила именования переменных

Основные правила именования:
- могут содержать буквы (заглавные буквы и строчные буквы латиницы, и что удивительно кириллицыб цифры, нижнее подчёркивание)
- не могут начинаться с цифры
- регистрозависимые
- не могут являться ключевыми словами Python

Проверить, является ли переменная ключевым словом можно следующим образом:


```python
import keyword
print(keyword.iskeyword("class"))  # True
print(keyword.iskeyword("my_var")) # False
```

Также не рекомендуется называть переменные названиями функций, так как в противном случае могут возникнуть ошибки, при вызове функции с аналогичным названием.

```python
min = "a"
print("min =", min)
print("Попытка выполнения функции:")
try:
    print(min(1, 3))
except TypeError as error:
    print(error)
```

**Существуют определённые стили для именования переменных (PEP8):**
1. snake_case (основной стиль для переменных и функций)
2. UPPER_CASE (для констант)
3. CamelCase  (для классов)
4. Защищенные имена (single underscore)
5. Специальные имена (двойные подчеркивания)
6. Имена для игнорирования (одинарное подчеркивание)


```python
# CamelCase
class ClassName():
    pass

#snake_case
ordinary_var = 10

# UPPER_CASE
CONSTANT_VAR = 15
```

```python
class Example:
    def __init__(self):
        self.public = 1      # Доступная переменная
        self._protected = 2  # Доступная с пометкой "не трогать"
        self.__private = 3   # "Недоступная переменная"

obj = Example()
print(obj.public)           # 1 (OK)
print(obj._protected)       # 2 (но не рекомендуется)
try:       
    print(obj.__private)        # Ошибка!
except AttributeError as error:
    print(error)

# Способ достать приватную переменную:
# Не использовать!!!
print(obj._Example__private)
```

### Область видимости

Область видимости - это, простыми словами часть кода, в которой переменная видна и доступна.

Существует **4** уровня видимости:
1. Локальная перемнная (local - ществует внутри функции)
2. Охватывающая (enclosing - во внешней функции)
3. Глобальная (globhal - глобальная на уровне модуля)
4. Встроенная (build-in - встроенные функции/типы)

Правило LEGB (поиск переменных):

Интерпретатор ищет переменную в порядке:
*Local* → *Enclosing* → *Global* → *Built-in*


Существуют следующие слова для работы с доступом к переменным:


> global

Доступ к глобальной переменной

```python
counter = 0

def increment():
    global counter
    counter += 1  # Меняем глобальную переменную
    print(counter)

increment()
```

>nonlocal

Доступ к переменной внешней функции

```python
def outer():
    x = 0
    
    def inner():
        nonlocal x
        x = 1  # Меняем x из outer()
    
    inner()
    print(x)

outer()
```

### Присваивание значений

Присваивание бывает трёх видов:
- обычное присваивание
- множественное присваивание
- присваивание распоковкой

```python
# Обычное присваивание
a = 10

# Множественное присванивание
a = b = c = 10
print(a, b, c)
a, b, c = 5, 10, 15
print(a, b, c)

# Меняем значения переменных местами, благодяря присваиванию
b, d = 5, 10
b, d = d, b
print(b, d)
```

```python
# Присваивание распаковкой
a, b, c = [1, 2, 3]
a, b, c = (1, 2, 3)
print(a, b, c)

a, *b, c, d = [1, 2, 3, 4, 5]
print(a, b, c, d)
```

### Удаление переменных


Удаление переменных происходит через оператор *"del"*

Что происходит:
- удаляет имя из пространства имён
- не удаляет объект из памяти напрямую
- счётчик ссылок объекта не уменьшается на 1
- если счётчик становится 0, позже объект удаляется "сборщиком мусора"

```python
import sys
x = [1, 2, 3]      # Создаём объект списка
y = x              # Ещё одна ссылка на тот же объект

print(sys.getrefcount(x) - 1)  #Количество ссылок (=2, от x и y)

del x              # Удалили имя x, но объект живёт!
print(y)           # [1, 2, 3] - объект доступен через y

del y              # Теперь удалили последнюю ссылку
# Счётчик ссылок = 0 => объект готов к удалению GC
```

GC - Garbage Collector (сборщик мусора) — это механизм Python, который автоматически освобождает память от неиспользуемых объектов.

### Особенности присваивания для разных типов

Иммутабельные типы (создается новый объект)

```python
x = 10
print(id(x))  # уникальный id

x = x + 5     # Создается НОВЫЙ объект
print(id(x))  # другой ID

# То же самое для строк
s = "hello"
print(id(s))  # некоторый ID

s = s + " world"  # Создается НОВАЯ строка
print(id(s))  # другой ID
```

Мутабельные типы (изменяется существующий объект)

```python
lst = [1, 2, 3]
print(id(lst))  # некоторый ID

lst.append(4)   # Изменяем существующий объект
print(id(lst))  # Тот же самый ID

lst = lst + [5] # Создается НОВЫЙ объект
print(id(lst))  # другой ID
```

### Проверка существования переменных

Проверить наличие переменной можно в двух облостях: локальной (внутри функции) и глобальной. 

```python
# Проверка в глобальной области
x = 10
print('x:', 'x' in globals()) 
print('y:', 'y' in globals())

# Проверка в локальной области
def my_func():
    global x
    local_var = "test"
    print('local_var:', 'local_var' in locals())  # True
    print('x в локальных:','x' in locals())          # False
    print('x в глобальных:','x' in globals())         # True

my_func()
```
