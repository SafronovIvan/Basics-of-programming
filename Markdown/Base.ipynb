# **Markdown**
Это небольшая шпоргалка, которуя я подготовил для того, чтобы в дальнейшем создавать лучшие комментарии и описания для своих работ.

Markdown - один из самых популярных языков разметки, он наполнен разнообразными функциями, которые будет весьма кстати упомянуть.

Я разделил этот файл на следующие разделы:
- абзацы и заголовки
- выделение текста
- списки
- вставка (ссылки, картинки, таблицы)

### **Абзацы и заголовки**

#### **Абзацы**

Если вы хотите сделать параграф, то нужно создать пустую строку после абзаца:
```
Первый абзац

Второй абзац
Продолжение второго абзаца
```
В противном случае текст будет по прежнему писаться в той-же строке

>Первый абзац
>
>Второй абзац
>Продолжение второго абзаца

Чтобы осуществить перенос строки - можно воспользоваться следующими способами:
```
Поставить в конце строки два или больше пробела "  "
Поставить в конце строки обратный слэш\
Пспользовать HTML-тег <br>
Последняя строка
```
>Перенос с помощью пробелов   
>Перенос с помощью обратного слэша\
>Перенос с помощью HTML-тега <br>
>Последняя строка

#### **Заголовки**

Если мы хотим выделить какую-то часть текста как заголовок (существуют уровни от 1 до 6), то нам помогут `#`:

```
# Заголовок первого уровня
## Заголовок второго уровня
### Заголовок третьего уровня
#### Заголовок четвёртого уровня
##### Заголовок пятого уровня
###### Заголовок шестого уровня
```
># Заголовок первого уровня
>## Заголовок второго уровня
>### Заголовок третьего уровня
>#### Заголовок четвёртого уровня
>##### Заголовок пятого уровня
>###### Заголовок шестого уровня

Для разделения абзауев на части или выделение заголовка чертой можно использовать `***`:
```
Текст 1
***
Текст 2
```
>Текст 1
>***
>Текст 2

### **Выделение текста**
Существует четыре основных способов выделения текста, это:
- *Курсив* (*)
- **Жирный** (**)
- ***Жирный курсив*** (***)
- ~~Зачёркнутый~~ (~~)

Для того чтобы выделить текст одним из вышеуказанных способов надо использовать соответствующий символ или комбинацию из символов по обе стороны от выделяемого текста.
```
*Курсив* (*)
**Жирный** (**)
***Жирный курсив*** (***)
~~Зачёркнутый~~ (~~)
```

Также есть вспомогательнве способы выделения текста
```
Прямое толкование
```
> Цитата

Чтобы их реализвать надо использовать следующие символы:

"`" - для прямого толкования по обе стороны от текста

"\>" \- для цитаты - для каждой строки кода с цитатой (можно создавать многоуровневые цитаты)
> Первый уровень цитаты
>> Второй уровень цитаты




### **Списки**

**Нумерованные списки.**

Нумерованные списки будут начинаться с первого числа и идти дальше по порядку, не взирая на указанный вами.

```
15. Первый пункт
3. Второй пункт
1. Третий пункт
```

>15. Первый пункт
>3. Второй пункт
>1. Третий пункт

**Ненумерованные списки.**

Для создания такого списка можетиспользоваться любой из символов: "+", "-", "*"
```
- Первый пункт
+ Второй пункт
* Третий пункт
```
>- Первый пункт
>+ Второй пункт
>* Третий пункт

**Вложенные списки**

Для создания вложенных списков следует использовать знак табуляции:
```
1. Пункт
    1. Подпункт
        1. Подподпункт

- Пункт
    - Подпункт
        - Подподпункт
```
>1. Пункт
>    1. Подпункт
>        1. Подподпункт

>- Пункт
>    - Подпункт
>        - Подподпункт



### **Вставка**

#### **Ссылки**

Для того чтобы просто вставить ссылку можно использовать угловые скобки:

```
<https://github.com/SafronovIvan/Basics-of-programming/blob/main/Markdown/Base.ipynb>
```
> <https://github.com/SafronovIvan/Basics-of-programming/blob/main/Markdown/Base.ipynb>

Если перед ссылкой в квадратных скобках указать ключевую фразу, то ссылка оформится как эта фраза. Однако саму ссылку теперь стоит указывать в круглых скобках.

```
[Ссылка на эту страницу](https://github.com/SafronovIvan/Basics-of-programming/blob/main/Markdown/Base.ipynb)
```
> [Ссылка на эту страницу](https://github.com/SafronovIvan/Basics-of-programming/blob/main/Markdown/Base.ipynb)

Если в круглых скобках после ссылку указать подсказку, то она всплывёт при наведении на ссылку. Однако подсказку стоит указывать в кавычках.

```
[Ссылка на эту страницу](https://github.com/SafronovIvan/Basics-of-programming/blob/main/Markdown/Base.ipynb "Подсказка")
```
> [Ссылка на эту страницу](https://github.com/SafronovIvan/Basics-of-programming/blob/main/Markdown/Base.ipynb "Подсказка")

Второй вариант указывания подсказки - справочный метод. Мы указываем сноску, а ниже в коде пишем для этой сноски ссылку:
```
[Ссылка на эту страницу]

[Ссылка на эту страницу]: https://github.com/SafronovIvan/Basics-of-programming/blob/main/Markdown/Base.ipynb
```
>[Ссылка на эту страницу]

>[Ссылка на эту страницу]: https://github.com/SafronovIvan/Basics-of-programming/blob/main/Markdown/Base.ipynb

#### **Таблицы**

Таблица создаётся с помощью символа "|", который используется как разграничитель между столбцами.
```
|Столбец 1|Столбец 2|Столбец 3|
|-|-|-|
|значение 1| значение 2 | значение 3|
```
|Столбец 1|Столбец 2|Столбец 3|
|-|-|-|
|значение 1| значение 2 | значение 3|


Для того чтобы выравнять текст в таблице надо применить ":":
```
|Столбец 1|Столбец 2|Столбец 3|
|:-|:-:|-:|
|Равнение по левому краю|Равнение по центру|Равнение по правому краю|
|Запись|Запись|Запись|
```
