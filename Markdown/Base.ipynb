# **Markdown**

Markdown - один из самых популярных языков разметки, он наполнен разнообразными функциями, которые будет весьма кстати упомянуть.

Я разделил этот файл на следующие разделы:
- абзацы и заголовки
- выделение текста

### **Абзацы и заголовки**

#### **Абзацы**

Если вы хотите сделать параграф, то нужно создать пустую строку после абзаца:
```
Первый абзац

Второй абзац
Продолжение второго абзаца
```
В противном случае текст будет по прежнему писаться в той-же строке

>Первый абзац
>
>Второй абзац
>Продолжение второго абзаца

Чтобы осуществить перенос строки - можно воспользоваться следующими способами:
```
Поставить в конце строки два или больше пробела "  "
Поставить в конце строки обратный слэш\
Пспользовать HTML-тег <br>
Последняя строка
```
>Перенос с помощью пробелов   
>Перенос с помощью обратного слэша\
>Перенос с помощью HTML-тега <br>
>Последняя строка

#### **Заголовки**

Если мы хотим выделить какую-то часть текста как заголовок (существуют уровни от 1 до 6), то нам помогут `#`:

```
# Заголовок первого уровня
## Заголовок второго уровня
### Заголовок третьего уровня
#### Заголовок четвёртого уровня
##### Заголовок пятого уровня
###### Заголовок шестого уровня
```
># Заголовок первого уровня
>## Заголовок второго уровня
>### Заголовок третьего уровня
>#### Заголовок четвёртого уровня
>##### Заголовок пятого уровня
>###### Заголовок шестого уровня

Для разделения абзауев на части или выделение заголовка чертой можно использовать `***`:
```
Текст 1
***
Текст 2
```
>Текст 1
>***
>Текст 2

### **Выделение текста**
Существует четыре основных способов выделения текста, это:
- *Курсив* (*)
- **Жирный** (**)
- ***Жирный курсив*** (***)
- ~~Зачёркнутый~~ (~~)

Для того чтобы выделить текст одним из вышеуказанных способов надо использовать соответствующий символ или комбинацию из символов по обе стороны от выделяемого текста
