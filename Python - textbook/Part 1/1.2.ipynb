### **Целые числа - int**


```python
# Прямое присваивание
x = 10          # десятичная
y = -5          # отрицательное
z = 0           # ноль
w = 1_000_000   # с разделителем

# Преобразование из других типов
a = int(3.14)       # 3 (отбрасываем дробную часть)
b = int("25")       # 25 (из строки)
c = int("1010", 2)  # 10 (из указанной системеы счистления)
d = int(True)       # 1 (True -> 1, False -> 0)
```

Для работы с различными системамми счистления существуют функции для ускоренного перевода.


Если мы хотим перевести из десятичной в двоичную, восьмеричную или шестнадцатеричную можно использовать соотвествующие функции - bin, oct, hex.

Для перевода в десятичную систему из другой системы счистления с иным основанием следует использовать функцию int для превращения строки в число с дополниительным параметром (системой счистления), или использовать префикасы систем счистления.

```python
# Перевод из десятичной системы счистления
bin_str = bin(10)   # '0b1010'
oct_str = oct(10)   # '0o12'
hex_str = hex(10)   # '0xa'

# Перевод в десетичную систему счистления
bin_num = 0b1010    # 10
oct_num = 0o12      # 10
hex_num = 0xA       # 10

# Произвольная система счисления (2-36)
num = int("FF", 16)  # 255
num = int("22", 3)   # 8
```

#### **Методы и атрибуты целых чисел**

##### Методы преобразования



> bit_length() - длина в битах\
to_bytes() - перевод в байты\
from_bytes() - перевод из байтов

```python
print((0).bit_length())     # 0
print((1).bit_length())     # 1
print((10).bit_length())    # 4 (1010 в двоичной)
print((-10).bit_length())   # 4 (для отрицательных — без знака)

# Преобразование int в байты
x = 1024
print(x.to_bytes(4, 'big'))     # b'\x04\x00'
print(x.to_bytes(4, 'little'))  # b'\x00\x04'
print(x.to_bytes(4, 'big'))     # b'\x00\x00\x04\x00'

# Для отрицательных чисел
y = -1024
print(y.to_bytes(4, 'big', signed=True))  # b'\xff\xff\xfc\x00'

# Из байтов
# Классовый метод
print(int.from_bytes(b'\x04\x00', 'big'))      # 1024
print(int.from_bytes(b'\x00\x04', 'little'))   # 1024
print(int.from_bytes(b'\xff\xff\xfc\x00', 'big', signed=True))  # -1024
```

little и big - порядок байтов (byte order) или эндианность (endianness), определяющая, как многобайтовые числа хранятся в памяти и передаются.

signed - показатель того, хранится ли число с указанием знака или без него

#### Операции с int

Математические операции


```python
print("Базовые")
print(10 + 3)   # 13
print(10 - 3)   # 7
print(10 * 3)   # 30
print(10 / 3)   # 3.333... (возвращает float!)

print("\nЦелочисленное деление")
print(10 // 3)  # 3
print(-10 // 3) # -4 (округляет вниз!)

print("\nОстаток от деления")
print(10 % 3)   # 1
print(-10 % 3)  # 2 (важно: результат всегда положительный)

print("\nВозведение в степень")
print(2 ** 3)   # 8
print(pow(2, 3)) # 8
print(pow(2, 3, 5)) # 3 (2^3 mod 5)
```

Побитовые операции

```python
x = 10  # 1010
y = 6   # 0110

# Поразрядовые операции
print(bin(x & y))        # 0b0010 = 2   (AND) 
print(bin(x | y))        # 0b1110 = 14  (OR)
print(bin(x ^ y))        # 0b1100 = 12  (XOR)

# Операции сдвига на бит
print(bin(x << 1))       # 0b10100 = 20
print(bin(x >> 1))       # 0b101 = 5
```

Сравнение

В коде ниже вы можете увидеть различные операции для сравнения, а именно:
- больше
- меньше
- равенство
- не равенство
- меньше или равно
- больше или равно

```python
a = (10 > 5)      # True
a = (10 < 5)      # False
a = (10 == 10)    # True
a = (10 != 5)     # True
a = (5 <= 10)     # True
a = (5 >= 10)     # False
```
