# Работа с Git

## 🚀 1. Начало работы

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git init` | Создаёт новый репозиторий | Создаёт скрытую папку `.git` — всё "магия" Git хранится там |
| `git status` | Показывает состояние | Что изменено, что добавлено в индекс, что готово к коммиту |
| `git log` | История коммитов | Показывает автора, дату, хеш коммита и сообщение |
| `git log --oneline --graph` | Визуальная история | Компактный вид + ASCII-граф ветвления |
| `git diff` | Неподтверждённые изменения | Показывает **конкретные строки**, которые вы изменили |
| `git diff main..feature` | Разница между ветками | Что изменилось в `feature` относительно `main` |


## ✏️ 2. Работа с коммитами (сохранение изменений)

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git add file.txt` | Добавить в индекс | Подготовить файл к коммиту (как поставить галочку) |
| `git add .` | Добавить всё | Добавляет **все** изменения в текущей папке и подпапках |
| `git commit -m "текст"` | Создать коммит | Сохраняет "снимок" индекса с сообщением/комментарием |
| `git commit --amend` | Изменить последний коммит | Позволяет добавить забытые файлы или изменить сообщение |
| `git reset HEAD~1` | Откатить локально | Удаляет последний коммит, но **сохраняет изменения** в рабочей папке |
| `git revert HEAD` | Безопасный откат | Создаёт **новый коммит**, который отменяет изменения старого (безопасно для командной работы) |


## 🌿 3. Ветки (branches)

**Что такое ветка?** Это просто подвижный указатель на определённый коммит. Позволяет вести параллельную разработку.

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git branch` | Список веток | Звёздочка `*` показывает текущую ветку |
| `git branch feature` | Создать ветку | Создаёт указатель на текущий коммит |
| `git branch -d feature` | Удалить ветку | Только если изменения уже слиты в другую ветку |
| `git branch -D feature` | Принудительно удалить | Удаляет ветку даже без слияния |
| `git branch -f main HEAD~3` | Переместить ветку | Перемещает указатель `main` на 3 коммита назад |
| `git checkout feature` | Переключиться | Меняет `HEAD` на ветку `feature` |
| `git checkout -b feature` | Создать + переключиться | Сокращение для двух команд |
| `git switch feature` | Переключиться (новый синтаксис) | Более понятная альтернатива `checkout` |
| `git switch -c feature` | Создать + переключиться | Новый синтаксис |


## 🔗 4. Объединение веток (слияние)

**Что такое слияние веток?** Это процесс объединения изменений из одной ветки в другую. Git позволяет вести параллельную разработку в разных ветках, а затем интегрировать результаты.

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git merge feature` | Влить ветку `feature` в текущую ветку | Git находит общего предка, сравнивает изменения и создаёт новый merge-коммит, у которого два родителя. Сохраняет полную историю ветвления, не переписывает коммиты |
| `git rebase feature` | Перебазировать текущую ветку на `feature` | Git берёт коммиты текущей ветки и "пересаживает" их на вершину ветки `feature`, создавая новые коммиты с новыми хешами. История становится линейной |
| `git rebase -i HEAD~4` | Интерактивный ребейз последних 4 коммитов | Открывает редактор, где можно изменять коммиты: менять сообщение, объединять, удалять, переупорядочивать. Позволяет очистить историю перед слиянием |

> ⚠️ **Золотое правило Git:** Никогда не делайте `rebase` на ветках, которые уже есть у других разработчиков. `merge` безопасен для общих веток, `rebase` — только для личных.


## 🧭 5. Перемещения по дереву коммитов

**Что такое HEAD?** Это специальный указатель в Git, который показывает, где вы сейчас находитесь. Если HEAD указывает на ветку — вы на последнем коммите этой ветки. Если HEAD указывает на конкретный коммит — это состояние называется "отделённый HEAD".

| Команда | Что делает | Теория |
|---------|------------|--------|
| `HEAD` | Указатель на текущий коммит или ветку | Показывает ваше текущее положение в дереве коммитов |
| `HEAD^` или `HEAD~1` | Первый родитель текущего коммита | Перемещает указатель на один коммит назад по истории |
| `HEAD^^` или `HEAD~2` | На 2 коммита выше | Перемещает указатель на два коммита назад |
| `git checkout main^` | Переключиться на родителя `main` | Переводит HEAD в состояние "отделённый HEAD", указывая на родительский коммит ветки `main` |
| `git checkout HEAD~3` | Переключиться на 3 коммита назад | Перемещает HEAD на три коммита назад от текущей позиции |
| `git branch -f main HEAD~3` | Переместить ветку | Принудительно перемещает указатель ветки `main` на указанное место (на 3 коммита выше), не переключаясь на неё |


## 📋 6. Копирование коммитов

**Что такое cherry-pick?** Это способ скопировать один или несколько коммитов из любого места в вашу текущую позицию. Создаются новые коммиты с новыми хешами, но с тем же содержимым.

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git cherry-pick abc123 def456` | Скопировать указанные коммиты | Берёт изменения из указанных коммитов и применяет их как новые коммиты в текущей позиции. Полезно для переноса конкретных исправлений без слияния целой ветки |
| `git rebase --onto main feature~3 feature` | Сложный ребейз | Переносит коммиты из ветки `feature`, начиная с третьего коммита от конца, на ветку `main`. Остальные коммиты остаются на месте. Позволяет перенести только часть ветки |


## 🏷️ 7. Теги (tags)

**Что такое теги?** Это постоянные метки на коммитах. В отличие от веток, они не перемещаются при создании новых коммитов. Используются для отметки релизов или важных точек в истории.

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git tag v1.0` | Создать лёгкий тег | Просто указатель на коммит, не хранит метаданных (автор, дата) |
| `git tag -a v1.0 -m "описание"` | Создать аннотированный тег | Хранит автора, дату, сообщение, может быть подписан GPG. Рекомендуется для официальных релизов |
| `git tag v0.5 abc123` | Присвоить тег конкретному коммиту | Создаёт тег не на текущем коммите, а на указанном (по хешу) |
| `git tag` | Показать все теги | Выводит список всех тегов в репозитории |
| `git tag -l "v1.*"` | Поиск тегов по шаблону | Показывает только теги, соответствующие шаблону (например, все теги v1.x) |
| `git tag -d v1.0` | Удалить тег локально | Удаляет тег только из вашего локального репозитория |
| `git push origin v1.0` | Отправить тег на удалённый репозиторий | Теги не отправляются автоматически при `git push`, их нужно отправлять отдельно |
| `git push origin --tags` | Отправить все теги | Отправляет все локальные теги на удалённый репозиторий |
| `git push origin --delete v1.0` | Удалить тег на удалённом репозитории | Удаляет тег с сервера |
| `git describe` | Показать ближайший тег | Выводит строку вида "v1.0-3-g67e1d82": ближайший тег, количество коммитов после него и хеш. Полезно для версионирования сборок |


## 🔍 8. Поиск ошибок — `git bisect`

**Что такое bisect?** Это бинарный поиск коммита, который ввёл баг. Позволяет быстро найти "плохой" коммит среди сотен и тысяч, деля диапазон поиска пополам на каждом шаге.

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git bisect start` | Начать бинарный поиск | Запускает режим bisect, подготавливая Git к поиску проблемного коммита |
| `git bisect bad` | Отметить текущий коммит как плохой | Указывает, что в текущем коммите присутствует баг |
| `git bisect good v1.0` | Отметить коммит как хороший | Указывает, что в этом коммите (по тегу или хешу) бага нет. Git автоматически переключается на средний коммит между good и bad |
| `git bisect reset` | Завершить поиск | Выходит из режима bisect и возвращает HEAD в исходное состояние |
| `git bisect run npm test` | Автоматический bisect | Git сам запускает указанную команду (например, тесты) на каждом коммите и определяет, хороший он или плохой. Коммит, где тест упадёт, будет виновным |


## 🌐 9. Удалённые репозитории

**Что такое удалённый репозиторий?** Это версия вашего проекта, хранящаяся на сервере (GitHub, GitLab, Bitbucket). `origin` — стандартное имя для удалённого репозитория. `origin/main` — локальная копия удалённой ветки main.

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git remote -v` | Показать подключённые удалённые репозитории | Выводит имена и URL всех удалённых репозиториев с указанием прав (fetch/push) |
| `git remote add origin https://...` | Добавить удалённый репозиторий | Привязывает локальный репозиторий к удалённому с именем `origin` |
| `git remote remove origin` | Удалить привязку | Удаляет связь с удалённым репозиторием, локальные изменения остаются |
| `git remote rename origin github` | Переименовать удалённый репозиторий | Меняет имя удалённого репозитория (например, с `origin` на `github`) |
| `git fetch` | Скачать изменения без слияния | Обновляет `origin/main` и другие удалённые ветки, но не трогает ваши локальные ветки |
| `git fetch origin main` | Скачать конкретную ветку | Обновляет только указанную удалённую ветку |
| `git pull` | Скачать и слить | Выполняет `git fetch`, затем `git merge` — объединяет удалённые изменения с вашими локальными |
| `git pull --rebase` | Скачать и сделать rebase | Выполняет `git fetch`, затем `git rebase` — переносит ваши локальные коммиты поверх удалённых, создавая линейную историю |
| `git push` | Отправить изменения | Загружает ваши локальные коммиты в удалённый репозиторий |
| `git push origin main` | Отправить ветку main в origin | Загружает только указанную ветку в удалённый репозиторий |
| `git push -u origin main` | Отправить и установить upstream | Отправляет изменения и запоминает связь между локальной и удалённой веткой. После этого можно просто `git push` |
| `git push origin --delete feature` | Удалить ветку на удалённом репозитории | Удаляет указанную ветку с сервера |
| `git push --all origin` | Отправить все ветки | Загружает все локальные ветки в удалённый репозиторий |



## 📡 10. Работа с удалёнными ветками

**Что такое удалённая ветка?** Это ветка, которая существует на сервере. Локально она доступна как `origin/название` и является "слепком" того, что было на сервере при последнем fetch.

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git branch -r` | Показать удалённые ветки | Выводит список всех веток, которые есть на удалённых репозиториях |
| `git branch -a` | Показать все ветки | Показывает и локальные, и удалённые ветки. Удалённые отмечены как `remotes/origin/...` |
| `git branch -vv` | Показать отслеживание веток | Выводит локальные ветки и показывает, какие удалённые ветки они отслеживают, а также разницу в коммитах (ahead/behind) |
| `git checkout -b feature origin/feature` | Создать локальную ветку из удалённой | Создаёт новую локальную ветку `feature` и настраивает её на отслеживание удалённой `origin/feature` |
| `git checkout --track origin/feature` | Создать ветку с отслеживанием | Сокращённый синтаксис: создаёт локальную ветку с тем же именем и настраивает отслеживание |
| `git checkout o/main` | Переключиться на указатель удалённой ветки | Переводит HEAD в состояние "отделённый HEAD", указывая на локальную копию удалённой ветки. `o/main` — сокращение для `origin/main` |
| `git branch -u origin/main` | Установить upstream для текущей ветки | Настраивает текущую ветку на отслеживание `origin/main`. После этого `git pull` и `git push` будут работать без указания ветки |
| `git diff main origin/main` | Сравнить локальную и удалённую ветки | Показывает различия между вашей локальной веткой и её удалённой копией |

## 🚨 11. Разрешение конфликтов

**Что такое конфликт?** Ситуация, когда Git не может автоматически объединить изменения, потому что две ветки изменили одни и те же строки в одном файле, или одна удалила файл, а другая его изменила.

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git merge --abort` | Прервать слияние | Отменяет процесс слияния и возвращает репозиторий в состояние до `git merge`. Используется, если конфликт слишком сложный |
| `git rebase --abort` | Прервать ребейз | Отменяет процесс ребейза и возвращает репозиторий в исходное состояние |
| `git rebase --continue` | Продолжить ребейз | После ручного разрешения конфликта и выполнения `git add` эта команда продолжает процесс ребейза |
| `git mergetool` | Запустить инструмент слияния | Открывает визуальный инструмент для разрешения конфликтов (например, VSCode, Meld, Beyond Compare) |
| `git checkout --ours file.txt` | Взять нашу версию файла | При конфликте выбирает версию файла из текущей ветки (HEAD), игнорируя изменения из вливаемой ветки |
| `git checkout --theirs file.txt` | Взять их версию файла | При конфликте выбирает версию файла из вливаемой ветки, игнорируя изменения в текущей ветке |
| `git add file.txt` | Отметить конфликт как решённый | После ручного исправления конфликта в файле, `git add` сообщает Git, что конфликт разрешён, и файл готов к коммиту |

> **Как разрешить конфликт вручную:** Откройте файл с конфликтом, найдите маркеры `<<<<<<<`, `=======`, `>>>>>>>`, выберите нужные строки (или объедините их), удалите маркеры, сохраните файл, выполните `git add` и затем `git commit` (или `git rebase --continue`).


## 💡 12. Полезные советы и скрытые команды

**Что такое reflog?** Это журнал всех перемещений HEAD в вашем локальном репозитории. Он сохраняет историю даже тех коммитов, которые вы "потеряли" после `reset` или `rebase`.

| Команда | Что делает | Теория |
|---------|------------|--------|
| `git stash` | Временно спрятать незакоммиченные изменения | Сохраняет текущие изменения в отдельное хранилище и возвращает рабочую папку к состоянию последнего коммида. Полезно, когда нужно срочно переключиться на другую задачу |
| `git stash pop` | Вернуть спрятанные изменения | Извлекает последние сохранённые изменения из stash и применяет их к рабочей папке. Затем удаляет их из stash |
| `git stash list` | Показать все сохранённые наборы | Выводит список всех stash-записей с их индексами и описаниями |
| `git stash drop` | Удалить последний stash | Удаляет последнюю запись из stash без применения изменений |
| `git stash apply` | Применить stash без удаления | Применяет изменения из stash, но оставляет их в stash для повторного использования |
| `git reflog` | Показать журнал HEAD | Выводит историю всех перемещений HEAD, включая коммиты, которые не видны в `git log`. Позволяет восстановить "потерянные" коммиты |
| `git reset --hard abc123` | Восстановить состояние по хешу | Перемещает HEAD на указанный коммит и полностью перезаписывает рабочую папку. Опасно, так как удаляет незакоммиченные изменения |
| `git clean -fd` | Удалить неотслеживаемые файлы | Удаляет все файлы и папки, которые не отслеживаются Git. `-f` — принудительно, `-d` — включая папки |
| `git blame file.txt` | Показать автора каждой строки | Для каждой строки в файле показывает, кто и в каком коммите её последний раз менял. Полезно для поиска, кто ввёл баг |
| `git grep "текст"` | Поиск по репозиторию | Ищет указанный текст во всех файлах репозитория (включая историю, если добавить флаг) |
| `git shortlog -sn` | Статистика авторов | Показывает список авторов и количество их коммитов, отсортированный по убыванию |


## 📝 13. Типичные сценарии работы

| Сценарий | Последовательность команд |
|----------|---------------------------|
| Начать новую фичу | Переключиться на main, обновить из удалённого репозитория, создать ветку фичи, опубликовать её на сервере |
| Обновить ветку из main через rebase | Переключиться на ветку фичи, скачать изменения из удалённого репозитория, выполнить rebase на origin/main, разрешить конфликты (если есть), принудительно отправить изменения (force-with-lease) |
| Отменить последний коммит и удалить изменения | Выполнить `git reset --hard HEAD~1` — это удалит коммит и все незакоммиченные изменения |
| Отменить последний коммит, сохранив изменения | Выполнить `git reset --soft HEAD~1` — коммит исчезнет, но изменения останутся в индексе |
| Временно переключиться на другую задачу | Спрятать текущие изменения через `git stash`, переключиться на другую ветку, поработать, вернуться обратно и применить stash |
| Исправить сообщение последнего коммита | Выполнить `git commit --amend -m "Новое сообщение"`, затем отправить с флагом `--force-with-lease` |
| Добавить забытый файл в последний коммит | Добавить файл через `git add`, затем выполнить `git commit --amend --no-edit`, отправить с `--force-with-lease` |
| Отменить уже отправленный push (безопасно) | Выполнить `git revert HEAD`, затем `git push`. Это создаст новый коммит, отменяющий изменения, не переписывая историю |
| Восстановить "потерянный" коммит после reset | Найти хеш коммита через `git reflog`, затем выполнить `git reset --hard [хеш]` |
| Удалить ветку локально и на сервере | Локально: `git branch -d feature`. На сервере: `git push origin --delete feature` |
